```python
import os
import shutil

def duplicate_files(base_dir: str, label_filenames: list):

    if not os.path.exists(base_dir):
        return f"Path does not exist - {base_dir}"

    labels_dir = os.path.join(base_dir, 'labels')
    images_dir = os.path.join(base_dir, 'images')

    if not os.path.isdir(labels_dir) or not os.path.isdir(images_dir):
        return f"Expected 'labels' and 'images' subfolders in {base_dir}, but one or both are missing or not directories."

    # --- Duplicate Label Files ---
    count_labels = 0
    for label_filename in label_filenames:
        src_path = os.path.join(labels_dir, label_filename)
        
        # Check if the source file exists before trying to copy
        if not os.path.exists(src_path):
            print(f"Warning: Label file not found at {src_path}")
            continue
            
        dest_filename = "dup1_" + label_filename
        dest_path = os.path.join(labels_dir, dest_filename)
        
        shutil.copy2(src_path, dest_path)
        count_labels += 1
    
    print(f"✅ Labels duplicated: {count_labels} files copied in {labels_dir}")

    # --- Duplicate Image Files ---
    count_images = 0
    for label_filename in label_filenames:
        # Assuming image files have a .jpg extension
        image_filename = label_filename.replace(".txt", ".jpg") 
        
        src_path = os.path.join(images_dir, image_filename)

        # Check if the source file exists before trying to copy
        if not os.path.exists(src_path):
            print(f"Warning: Image file not found at {src_path}")
            continue

        dest_filename = "dup1_" + image_filename
        dest_path = os.path.join(images_dir, dest_filename)
        
        shutil.copy2(src_path, dest_path)
        count_images += 1
        
    print(f"✅ Images duplicated: {count_images} files copied in {images_dir}")
    
    return "Done duplicating files."
```
